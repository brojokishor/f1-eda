```python
import pandas as pd
import numpy as np
```

```python
laps = pd.read_csv('../data/raw/monza_2023_laps.csv')
print(laps.shape)
laps.head()
```

```python
laps = laps[laps['IsAccurate'] == True]
print(laps.shape)
```

```python
laps = laps[laps['Deleted'] == False]
print(laps.shape)
```

```python
laps = laps.dropna(subset=['LapTime', 'Position', 'Sector1Time', 'Sector2Time', 'Sector3Time'])
print(laps.shape)
```

```python
time_cols = ['LapTime', 'Sector1Time', 'Sector2Time', 'Sector3Time']

for col in time_cols:
    laps[col] = pd.to_timedelta(laps[col]).dt.total_seconds()

laps[time_cols].head()
```

```python
key_cols = ['Driver', 'LapTime', 'Position', 'Compound', 'TyreLife', 'Team']
print(laps[key_cols].isnull().sum())
```

```python
laps.to_csv('../data/processed/monza_2023_laps_clean.csv', index=False)
print("Saved successfully")
print(f"Final shape: {laps.shape}")
```
